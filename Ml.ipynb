```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('Data.csv')
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Humidity (%)'].value_counts()
```

```python
df['Humidity (%)'].fillna(df['Humidity (%)'].mean(), inplace=True)
```

```python
df.info()
```

```python
df['Temperature (°C)'].fillna(df['Temperature (°C)'].mean(), inplace=True)
df.info()
```

```python
df['Pressure (hPa)'].fillna(df['Pressure (hPa)'].mean(), inplace=True)
df['Light Intensity (Lux)'].fillna(df['Light Intensity (Lux)'].mean(), inplace=True)
# df['CO2 (ppm)'].fillna(df['CO2 (ppm)'].mean(), inplace=True)
df['PM1.0 (µg/m³)'].fillna(df['PM1.0 (µg/m³)'].mean(), inplace=True)
df['PM2.5 (µg/m³)'].fillna(df['PM2.5 (µg/m³)'].mean(), inplace=True)
# df['PM10 (µg/m³)'].fillna(df['PM10 (µg/m³)'].mean(), inplace=True)
df.info()
```

```python

df['CO2 (ppm)'] = pd.to_numeric(df['CO2 (ppm)'], errors='coerce')
df['CO2 (ppm)'].fillna(df['CO2 (ppm)'].mean(), inplace=True)
df.info()
```

```python
df['PM10.0 (µg/m³)'].fillna(df['PM10.0 (µg/m³)'].mean(), inplace=True)
df.info()
```

```python
df['Uptime'].fillna(0, inplace=True)
```

```python
df['Sensors Online'].fillna(0, inplace=True)
```

```python
from IPython.display import FileLink

# save preprocessed dataframe to CSV in notebook working dir
df.to_csv('prdtreprocessed_data.csv', index=False)

# if you want a quick download link in Jupyter
FileLink('prdtreprocessed_data.csv')
```
